```python
# ============================================================
# NLP-BASED KEYWORD EXTRACTION SYSTEM
# Using TF-IDF + Gradio
# ============================================================

!pip install -q gradio scikit-learn

import gradio as gr
from sklearn.feature_extraction.text import TfidfVectorizer


# ------------------------------------------------------------
# KEYWORD EXTRACTION FUNCTION
# ------------------------------------------------------------

def extract_keywords(text):

    if not text or not text.strip():
        return (
            "Please enter some text.",
            "0",
            "No text was provided.",
            "Waiting for input..."
        )

    try:

        # Create TF-IDF model
        vectorizer = TfidfVectorizer(
            stop_words="english"
        )

        # Convert text into TF-IDF values
        matrix = vectorizer.fit_transform([text])

        # Get words
        words = vectorizer.get_feature_names_out()

        # Get scores
        scores = matrix.toarray()[0]

        # Sort words according to TF-IDF score
        ranked_words = sorted(
            zip(words, scores),
            key=lambda x: x[1],
            reverse=True
        )

        # Select top 10 keywords
        top_keywords = ranked_words[:10]

        # Format keywords
        keyword_text = "\n".join(
            [
                f"{i+1}. {word} — Score: {score:.3f}"
                for i, (word, score) in enumerate(top_keywords)
            ]
        )

        analysis = (
            "The system analyzed the input text using "
            "TF-IDF and ranked words according to their "
            "importance in the given text."
        )

        result = "Keywords extracted successfully."

        return (
            keyword_text,
            str(len(top_keywords)),
            analysis,
            result
        )

    except Exception as e:

        return (
            "Unable to extract keywords.",
            "0",
            str(e),
            "Please enter more text."
        )


# ------------------------------------------------------------
# CUSTOM CSS
# ------------------------------------------------------------

css = """
body {
    background-color: #0b0f19 !important;
}

.gradio-container {
    background-color: #0b0f19 !important;
    color: white !important;
    max-width: 1200px !important;
}

h1 {
    color: white !important;
    font-size: 34px !important;
}

h2, h3, p, label {
    color: white !important;
}

textarea, input {
    background-color: #374151 !important;
    color: white !important;
}

button {
    background-color: #5146e5 !important;
    color: white !important;
    font-weight: bold !important;
}
"""


# ------------------------------------------------------------
# WEB APPLICATION
# ------------------------------------------------------------

with gr.Blocks(
    theme=gr.themes.Base(),
    css=css,
    title="NLP-Based Keyword Extraction System"
) as demo:

    gr.Markdown(
        """
        # 🔑 NLP-Based Keyword Extraction System

        ### Extract the most important keywords from text

        This application uses **TF-IDF** to identify and rank
        important words in the given text.
        """
    )

    gr.Markdown(
        """
        **Features:**

        - 📝 Text analysis
        - 🔑 Keyword extraction
        - 📊 TF-IDF scoring
        - 🧠 NLP-based analysis
        """
    )

    with gr.Row():

        # LEFT SIDE
        with gr.Column():

            gr.Markdown("### 📝 Enter Text")

            text_input = gr.Textbox(
                label="Input Text",
                placeholder=(
                    "Example: Artificial intelligence and "
                    "machine learning are important technologies..."
                ),
                lines=10
            )

            extract_button = gr.Button(
                "🔑 Extract Keywords",
                variant="primary"
            )

        # RIGHT SIDE
        with gr.Column():

            gr.Markdown("### 🔎 Extraction Result")

            keyword_output = gr.Textbox(
                label="Important Keywords",
                lines=10,
                interactive=False
            )

            count_output = gr.Textbox(
                label="Number of Keywords",
                interactive=False
            )

            analysis_output = gr.Textbox(
                label="Analysis",
                lines=4,
                interactive=False
            )

            result_output = gr.Textbox(
                label="Result",
                interactive=False
            )

    # Example
    gr.Markdown("### 💡 Example Text")

    gr.Examples(
        examples=[
            [
                "Artificial intelligence is transforming "
                "healthcare, education, business and technology. "
                "Machine learning helps computers learn from data "
                "and make intelligent predictions."
            ]
        ],
        inputs=text_input
    )

    # Button action
    extract_button.click(
        fn=extract_keywords,
        inputs=text_input,
        outputs=[
            keyword_output,
            count_output,
            analysis_output,
            result_output
        ]
    )


# ------------------------------------------------------------
# LAUNCH
# ------------------------------------------------------------

demo.launch(share=True)
```
